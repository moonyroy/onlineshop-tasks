# بخش اول: تعیین اهداف کسب‌وکار


## سوال ۱

«مسئلهٔ اصلی کسب‌وکار را با زبان ساده توضیح دهید.»

وقتی مسئله را خواندم، اول از خودم پرسیدم شرکت دقیقاً چه چیزی را می‌خواهد زودتر بفهمد؟ برداشت من این بود که بعضی مشتری‌ها قبل از قطع خرید، نشانه‌هایی از فاصله‌گرفتن دارند و بعضی‌ها هم ظاهراً یک‌دفعه ناپدید می‌شوند. هدف این است که از سابقهٔ خرید یک فهرست قابل‌پیگیری بسازم، نه اینکه فقط چند خوشه یا یک مدل تحویل بدهم.

البته هر وقفهٔ خرید ریزش نیست. مشتری‌ای که معمولاً دیر‌به‌دیر خرید می‌کند، با کسی که عادت خرید پرتکرارش شکسته فرق دارد. همین تفاوت باید در تعریف مسئله دیده شود.


## سوال ۲

«اگر این پروژه اجرا نشود چه مشکلاتی برای شرکت ایجاد خواهد شد؟»

در سناریوی شرکت واقعی، اگر این کار انجام نشود ممکن است مشتری‌های در حال فاصله‌گرفتن دیر شناسایی شوند و منابع پیگیری هم به‌صورت یکسان بین همه تقسیم شود. در نتیجه، هم فرصت حفظ بعضی مشتری‌ها از دست می‌رود و هم ممکن است برای کسانی هزینه شود که اصلاً نیاز به پیشنهاد ندارند.

اما من اطلاعاتی از روش فعلی بازاریابی یا هزینه‌های این شرکت ندارم. پس این‌ها پیامدهای احتمالیِ سناریو هستند نه مشکلاتی که در داده ثابت کرده باشم.


## سوال ۳

«چه ذی‌نفعانی در این پروژه حضور دارند و هر کدام چه نیازی دارند؟»

اگر این پروژه در یک شرکت واقعی اجرا شود، به نظرم مدیریت نتیجهٔ کلی و ارزش تصمیم را می‌خواهد بازاریابی و CRM فهرست مشتری، دلیل هشدار و پیشنهاد اقدام را لازم دارند فروش هم برای مشتری‌های عمده به اولویت تماس نیاز دارد. مالی باید هزینه و حاشیهٔ سود را روشن کند و تیم داده و IT مسئول اجرای درست و نگهداری فرایند باشند.

خدمات مشتری می‌تواند دربارهٔ شکایت‌ها کمک کند و مسئول حریم خصوصی باید شرایط استفاده از داده و تماس را بررسی کند. خود مشتری هم ذی‌نفع است: پیگیری نباید به تماس اضافی و آزاردهنده تبدیل شود. در پروژهٔ من، این‌ها نقش‌های پیشنهادی‌اند و چنین تیمی واقعاً در اختیارم نیست.


## سوال ۴

«هدف اصلی کسب‌وکار را در یک جمله بنویسید.»

هدف کسب‌وکار این است که مشتری‌های در حال فاصله‌گرفتن زودتر شناسایی شوند و منابع محدود حفظ مشتری، با توجه به خطر و ارزش رابطه، بهتر تخصیص پیدا کنند.


## سوال ۵

«حداقل سه هدف فرعی برای این پروژه تعریف کنید.»

اول می‌خواهم تعریف روشن و قابل‌بررسی از توقف و افت خرید داشته باشم. دوم، تفاوت تیپ‌های مشتری را بشناسم تا همه را با یک الگو نسنجیم. سوم، یک امتیاز و پنج سطح خطر همراه با توضیح قابل‌فهم بسازم.

بعد، در حدی که داده اجازه بدهد، پیشنهاد محصول و برآورد ارزش آینده را برای تکمیل اولویت پیگیری اضافه می‌کنم. مستندسازی و تکرارپذیری هم هدف فرعی مهمی است، ولی اجرای واقعی کمپین و پایش زنده جزو تحویل این تمرین نیست.


## سوال ۶

«چه شاخص‌هایی نشان می‌دهد پروژه موفق بوده است و مدیریت چگونه آن را اندازه می‌گیرد؟»

من موفقیت را در دو سطح می‌بینم. در همین تمرین باید نشان بدهم مدل روی دادهٔ کنارگذاشته‌شده، مشتری‌های در معرض خطر را بهتر از یک روش ساده در بالای فهرست قرار می‌دهد و گروه‌ها و هشدارها قابل‌توضیح‌اند.

در شرکت واقعی، مدیریت باید اثر اقدام را با افزایش حفظ مشتری و سود افزایشی پس از هزینه‌ها بسنجد ترجیحاً با مقایسهٔ گروه اقدام و کنترل تصادفی. چون من دادهٔ کمپین ندارم، فعلاً نمی‌توانم این نتیجهٔ کسب‌وکاری را گزارش کنم.


## سوال ۷

«موفقیت از دید تیم بازاریابی و CRM چگونه اندازه‌گیری می‌شود و اگر مدل دقیق باشد اما به آن اعتماد نکنند، پروژه موفق است؟»

از نگاه CRM، یک خروجی خوب باید به‌موقع، قابل‌فهم و متناسب با ظرفیت پیگیری باشد. مثلاً معلوم باشد چرا این مشتری بالای فهرست آمده و چه کاری می‌شود برایش بررسی کرد.

اگر مدل از نظر آماری خوب باشد ولی کسی نتواند از آن استفاده کند، کاربردش ناقص است. در تمرین من می‌توانم وضوح خروجی و امکان استفاده از آن را بررسی کنم، ولی پذیرش واقعی تیم را بدون حضور آن تیم اندازه نمی‌گیرم.


## سوال ۸

«مزایای مالی، عملیاتی و رابطه‌ای احتمالی پروژه چیست و ارزش یک مشتری حفظ‌شده چقدر است؟»

فایدهٔ احتمالی مالی، حفظ خریدهای آینده و کم‌شدن هزینه‌های بی‌هدف است. از نظر عملیاتی، اولویت تماس روشن‌تر می‌شود و از نظر رابطه با مشتری هم می‌توان پیشنهاد مرتبط‌تری داد.

ولی ارزش مشتری را نباید مساوی آخرین فاکتورش بگیرم. برآورد درآمد آینده می‌تواند یک تقریب باشد، اما بدون حاشیهٔ سود، هزینهٔ خدمت و هزینهٔ اقدام نمی‌توانم ارزش سودی واقعی یا سود حفظ مشتری را دقیق حساب کنم. این اطلاعات در فایل فعلی نیست.


## سوال ۹

«این پروژه چه تصمیم‌های مشخصی را تغییر می‌دهد و مالک هر تصمیم کیست؟»

تصمیم‌هایی که می‌خواهم پشتیبانی کنم این‌ها هستند: چه کسی زودتر بررسی شود، دلیل این اولویت چیست و چه نوع پیشنهادی برایش مرتبط‌تر است. تصمیم قطعی دربارهٔ کانال تماس، زمان اجرا و تخفیف به اطلاعاتی نیاز دارد که در این فایل ندارم.

در سناریوی واقعی، بازاریابی و CRM مالک اقدام‌اند، فروش پیگیری حساب‌های مهم را انجام می‌دهد و مالی سقف هزینه را تعیین می‌کند. من خروجی تحلیلی و پیشنهاد می‌دهم مدل به‌تنهایی صاحب تصمیم نیست.


## سوال ۱۰

«یک بیانیهٔ رسمی هدف کسب‌وکار بنویسید.»

«هدف پیشنهادی کسب‌وکار، شناسایی زودهنگام مشتری‌های در حال فاصله‌گرفتن و اولویت‌بندی اقدام‌های حفظ مشتری بر اساس خطر، ارزش موردانتظار و محدودیت منابع است.»

برای هدف عددیِ کاهش ریزش یا بازگشت سرمایه، باید هزینه‌ها، ظرفیت اجرا و نتایج کمپین موجود باشد. در این تمرین این‌ها مشخص نیستند، پس عدد قطعی نمی‌گذارم.


# بخش دوم: ارزیابی وضعیت


## سوال ۱۱

«چه اطلاعاتی را قبل از شروع پروژه باید از شرکت دریافت کنید؟»

اول از همه، تعریف فعلی شرکت از ریزش، تصمیم موردانتظار و ظرفیت پیگیری را می‌پرسیدم. بعد سراغ تاریخچهٔ خرید، مرجوعی، مشتری، محصول و دسته‌بندی کالا می‌رفتم. حاشیهٔ سود، سابقهٔ کمپین و پاسخ مشتری، کانال و مجوز تماس و تیکت‌های خدمات هم تحلیل را کاربردی‌تر می‌کردند.

در عمل فقط فایل تراکنش را دارم. بنابراین این فهرست، نیاز اطلاعاتیِ سناریوی واقعی است و نباید طوری نوشته شود که انگار قرار است همهٔ این داده‌ها به من تحویل داده شوند.


## سوال ۱۲

«چه سوال‌های کلیدی باید از مدیریت، بازاریابی، CRM و مدیران حساب پرسید؟»

از مدیریت می‌پرسیدم: خروجی قرار است چه تصمیمی را تغییر بدهد و موفقیت از نظر شما چیست؟ از بازاریابی می‌پرسیدم: چند مشتری را می‌توانید پیگیری کنید و چه پیشنهادهایی قابل‌اجراست؟ از CRM می‌پرسیدم: سابقهٔ تماس و امکان اندازه‌گیری نتیجه داریم؟

از مسئول مشتری‌های عمده هم دربارهٔ چرخهٔ خرید و نشانه‌های فاصله‌گرفتن سؤال می‌کردم. این سؤال‌ها برای روشن‌کردن مجهولات‌اند من فعلاً پاسخی از چنین جلسه‌ای ندارم.


## سوال ۱۳

«محدودیت‌های عمومی، مالی و فنی پروژه چیست؟»

محدودیت واقعی من، تاریخچهٔ حدود یک‌ساله، شناسهٔ خالی در بخشی از رکوردها، نبود برچسب آمادهٔ ریزش و نبود اطلاعات کمپین و سود است. زمان و امکانات دوره هم باید در اندازهٔ پروژه لحاظ شوند، ولی مهلت دقیق هنوز در این سند تعیین نشده است.

در سناریوی شرکت، بودجهٔ تماس و تخفیف هم محدودیت مهمی است. چون عدد آن را ندارم، ظرفیت‌های فرضی را فقط به‌عنوان سناریو بررسی می‌کنم. اجرای کمپین و استقرار را هم خارج از تحویل فعلی نگه می‌دارم.


## سوال ۱۴

«محدودیت‌های قانونی، حریم خصوصی و اخلاقی چیست؟»

برای این تمرین از شناسه‌ها فقط در حد تحلیل و تحویل آموزشی استفاده می‌کنم و اطلاعات شناسایی اضافی منتشر نمی‌کنم. دادهٔ تراکنش مجوز تماس واقعی با مشتری به من نمی‌دهد.

اگر پروژه در شرکت اجرا شود، مبنای استفاده از داده، سطح دسترسی، مدت نگهداری و شرایط تماس باید با مسئول مربوط بررسی شود. اینجا حکم حقوقی دربارهٔ قانون مشخصی نمی‌دهم. از نظر اخلاقی هم پیشنهاد نباید فریبنده یا آزاردهنده باشد و تصمیم مهم باید قابل‌توضیح و قابل‌بازبینی بماند.


## سوال ۱۵

«مهم‌ترین ریسک‌های پروژه چیست و احتمال و شدت هر کدام چقدر است؟»

بزرگ‌ترین ریسک برایم تعریف اشتباه ریزش است: احتمال آن را فعلاً متوسط و اثرش را زیاد در نظر می‌گیرم. سوگیری ناشی از شناسهٔ خالی را احتمال بالا و اثر متوسط تا زیاد، و ناپایداری مدل یا خوشه‌ها را احتمال متوسط و اثر متوسط می‌گذارم.

این درجه‌ها قضاوت اولیه‌اند، نه احتمال محاسبه‌شده. برای اولی تحلیل حساسیت تعریف‌ها، برای دومی جداسازی درست داده و برای بقیه بررسی پوشش و پایداری لازم است. در اجرای واقعی، بی‌اثر بودن اقدام و استفاده‌نشدن از خروجی هم ریسک‌های مهمی هستند در این تمرین اثر اقدام قابل‌سنجش نیست.


## سوال ۱۶

«چه داده‌هایی لازم است، کدام داده‌ها سخت به دست می‌آید و اگر داده کافی یا درست نباشد چه می‌کنید؟»

برای هستهٔ تحلیل، تراکنش معتبر و شناسهٔ مشتری ضروری است. دسته‌بندی محصول، سود، سابقهٔ اقدام و پاسخ مشتری کمک می‌کنند پیشنهادها و تحلیل اقتصادی بهتر شوند، اما برای شروع همهٔ آن‌ها را ندارم.

اطلاعات سود و کمپین معمولاً نیاز به منبع جدا دارند. اگر به آن‌ها دسترسی نباشد، دامنهٔ ادعا را کوچک می‌کنم مثلاً درآمد آینده را به‌جای سود گزارش می‌دهم و اثر حفظ مشتری را قطعی نمی‌دانم. اگر کیفیت داده برای یک خروجی کافی نبود، آن بخش را محدود یا متوقف می‌کنم و دلیلش را می‌نویسم.


## سوال ۱۷

«تحلیل SWOT پروژه را به صورت روایی ارائه دهید.»

قوت پروژه این است که سابقهٔ تراکنش داریم و خروجی موردنیاز هم نسبتاً روشن است: فهرست اولویت‌دار مشتری‌ها. ضعف آن، کوتاهی تاریخچه و نبود اطلاعات سود، کمپین و شناسه برای بخشی از رکوردهاست.

فرصتش این است که از یک تحلیل توصیفی به یک پیشنهاد قابل‌پیگیری برسیم و مسیر را تکرارپذیر کنیم. تهدیدش هم این است که با تعریف نادرست ریزش یا ارزیابی نامناسب، یک نتیجهٔ ظاهراً عالی اما کم‌فایده بسازیم. پشتیبانی واقعی مدیریت را هم چون شواهدی ندارم، به‌عنوان قوت تأییدشده نمی‌نویسم.


## سوال ۱۸

«دقیقاً چه چیزی را ریزش می‌دانیم، آیا کاهش تدریجی هم ریزش است، مشتری فعال کیست و افق زمانی چیست؟»

اول باید مشتریِ هدف پیشگیری را از مشتری غیر فعال جدا کنم. پیشنهاد اولیه‌ام این است که مشتری قابل‌امتیازدهی حداقل یک خرید معتبر در ۱۲۰ روز آخر بازهٔ مشاهده داشته باشد. این عدد فعلاً فرض است باید با فاصله‌های خرید و گزینه‌های ۹۰ و ۱۸۰ روز بررسی شود. مشتری بیرون این مجموعه را جدا گزارش می‌کنم، نه اینکه از کل خروجی حذف شود. برای مشتری تازه‌وارد و قدیمی هم مدت رابطه را بررسی می‌کنم.

ریزش سخت را نداشتن خرید معتبر در بازهٔ بعدی تعریف می‌کنم. برای ریزش نرم، افت قابل‌توجه نرخ تعداد خرید یا مبلغ خرید را بررسی می‌کنم مثلاً ۴۰ درصد به‌عنوان نقطهٔ شروع. چون بازه‌ها هم‌طول نیستند، مجموع خرید را مستقیم مقایسه نمی‌کنم و مدت حضور واقعی مشتری هم مهم است.

یک نشانهٔ دیگر، طولانی‌شدن فاصله از آخرین خرید نسبت به میانهٔ فاصلهٔ قبلی است مثلاً بیش از دو برابر، فقط برای مشتری با سابقهٔ کافی مثل حداقل سه فاکتور. این فاصله در پایان بازهٔ بعدی فقط برای برچسب استفاده می‌شود. آستانه‌های ۳۰/۴۰/۵۰ درصد و ضریب‌های ۱٫۵/۲/۳ را بررسی می‌کنم و می‌سنجم ترکیب شرط‌ها بیش از حد برچسب مثبت نسازد.

پیشنهاد زمانی، مشاهده تا پایان اوت و بررسی رفتار سپتامبر تا ۹ دسامبر ۲۰۱۱ است. نبود خرید در این دوره به معنی رفتن همیشگی نیست مشتری‌های با چرخهٔ بلند ممکن است هنوز فرصت کافی برای خرید نداشته باشند و باید با عدم‌اطمینان جدا مشخص شوند.


## سوال ۱۹

«پنجرهٔ مشاهده و پنجرهٔ عملکرد چیست و چگونه اطلاعات هر بازه را جدا نگه می‌داریم؟»

پنجرهٔ مشاهده یعنی گذشته‌ای که هنگام امتیازدهی در اختیار دارم پیشنهاد من ۱ دسامبر ۲۰۱۰ تا پایان ۳۱ اوت ۲۰۱۱ است. پنجرهٔ عملکرد یعنی بازه‌ای که رفتار بعدی را می‌بینم از ۱ سپتامبر تا ۹ دسامبر ۲۰۱۱، حدود صد روز.

ویژگی، خوشه، قواعد و مدل خرید فقط باید اطلاعات قبل از زمان مرجع را ببینند. آینده برای برچسب و ارزیابی است. اما این جداسازی به‌تنهایی آزمون مستقل نمی‌سازد: مشتری‌ها یا نمونه‌های زمانی آموزش، تنظیم، کالیبراسیون و آزمون باید مناسب جدا شوند و تبدیل‌های قابل‌آموزش فقط روی آموزش برازش شوند.

اگر نقطهٔ برش را تغییر بدهم، افق بررسی را تا جای ممکن ثابت می‌گیرم. محدودیت تاریخچه ممکن است تعداد آزمون‌ها را کم کند. ارزیابی تاریخی را هم از امتیازدهی در آخرین تاریخ فایل جدا می‌دانم برای فهرست آخر فایل، آیندهٔ واقعی در دسترس نیست.


## سوال ۲۰

«مشتریان عمده و خرده‌فروشی چه تفاوتی دارند و یک‌بار خریداران و فصل چهارم را چگونه مدیریت می‌کنیم؟»

مشتری حجیم ممکن است از نظر مبلغ و فاصلهٔ خرید با بقیه فرق داشته باشد پس خرید بزرگ را خودکار خطا نمی‌دانم. با اندازهٔ سبد، حجم و تکرار خرید می‌توانم یک پرچم احتمالی بسازم و اگر تفاوت جدی بود، جدا تحلیل کنم. از این پرچم هویت قطعی عمده‌فروش نتیجه نمی‌گیرم.

مشتری تک‌خریدی فاصلهٔ خرید قابل‌اعتمادی ندارد و مسئلهٔ خرید دوم برایش مهم است. با این حال می‌تواند با صفر خرید تکراری در BG/NBD حضور داشته باشد محدودیت سابقهٔ تکراری مثبت برای بخش پولی Gamma-Gamma جداست. سهم این مشتری‌ها را باید محاسبه کنم و عددی از پیش قطعی ندارم.

برای فصل چهارم هم تاریخچهٔ چندساله ندارم. مقایسه با تغییر گروه مشابه و جابه‌جایی نقطهٔ برش فقط بررسی حساسیت‌اند، نه راه‌حل قطعی فصل. این محدودیت در نتیجه باقی می‌ماند.


## سوال ۲۱

«با توجه به نبود شناسهٔ مشتری در بخشی از داده‌ها، واحد تحلیل چیست و چقدر می‌توان به شناسه اعتماد کرد؟»

واحد تحلیل اصلی من مشتری با CustomerID است، ولی واحد رکورد خام یک قلم فاکتور است. برای تعداد خرید باید فاکتور مستقل را بشمارم، نه ردیف کالا را.

تراکنش ناشناس را نمی‌توان به تاریخچهٔ مشتری وصل کرد بنابراین در مدل ریزش فردی وارد نمی‌شود، ولی می‌تواند برای شناخت پوشش داده و زمینهٔ فروش بررسی شود. با این حال باید ببینم رکوردهای ناشناس چه تفاوتی دارند تا محدودیت تعمیم را روشن کنم. شناسه هم یک شناسهٔ ثبت‌شده است و به‌تنهایی ثابت نمی‌کند هر شخص یا سازمان دقیقاً یک حساب دارد.


## سوال ۲۲

«برای هر سطح ریسک چه اقدامی، توسط چه کسی، از چه کانالی و با چه بودجه‌ای انجام می‌شود؟»

به نظرم پنج سطح خطر به‌تنهایی برای تعیین اقدام کافی نیست ارزش و تیپ مشتری را هم کنار آن می‌گذارم. برای خطر خیلی زیاد و ارزش بالا، بررسی شخصی و تماس مسئول حساب را پیشنهاد می‌دهم برای خطر بالا، یادآوری هدفمند و پیشنهاد مرتبط برای خطر متوسط، پیگیری سبک‌تر و برای خطر کم یا خیلی کم، حفظ رابطه بدون تخفیف بی‌دلیل.

این‌ها فرضیهٔ اقدام‌اند، نه نسخهٔ اثبات‌شده. در سناریوی واقعی، CRM و بازاریابی مالک اجرا و فروش مسئول حساب‌های مهم است. کانال به دسترسی و مجوز تماس و بودجه به هزینه، سود و اثر اقدام بستگی دارد هیچ‌کدام را از فایل تراکنش به‌تنهایی تعیین نمی‌کنم.


## سوال ۲۳

«چه چیزی باعث می‌شود مدل مورد اعتماد و استفاده قرار گیرد و چه دادهٔ اضافه‌ای تصمیم را بهتر می‌کند؟»

اول باید توضیح بدهم چرا مشتری پرخطر شده، بعد نشان بدهم رتبه‌بندی روی دادهٔ دیده‌نشده بهتر از روش ساده بوده است. خروجی هم باید با ظرفیت پیگیری سازگار و قابل‌استفاده باشد. در محیط واقعی، یک آزمایش کوچک و بازخورد استفاده‌کننده کمک می‌کند.

حاشیهٔ سود، سابقهٔ پیشنهاد و پاسخ، تیکت خدمات، دسته‌بندی محصول و کانال تماس تصمیم را بهتر می‌کنند. آزمون مدل روی گذشته را با گروه کنترل کمپین یکی نمی‌گیرم اولی کیفیت پیش‌بینی را می‌سنجد و دومی اثر اقدام را.


## سوال ۲۴

«چه فرضیه‌های روان‌شناختی دربارهٔ فاصله‌گرفتن و پاسخ به نگهداشت باید آزمایش کنیم و مرزهای اخلاقی کجاست؟»

حدس اولیه‌ام این است که بزرگ‌شدن فاصلهٔ خرید، کوچک‌شدن سبد یا کم‌شدن تنوع ممکن است قبل از توقف خرید دیده شود. این بخش را می‌توانم با تاریخچهٔ تراکنش بررسی کنم.

اما حدس‌هایی مثل اثر یادآوری، تخفیف مکرر، تماس زیاد یا حس بی‌عدالتی به اطلاعات اقدام و آزمایش نیاز دارند. از این فایل نمی‌توانم انگیزه یا وضعیت روانی مشتری را تشخیص بدهم. در سناریوی واقعی هم نباید از پیام فریبنده، فشار یا تماس ناخواسته استفاده شود فرضیهٔ روان‌شناختی را به نتیجهٔ قطعی تبدیل نمی‌کنم.


## سوال ۲۵

«چه فرض‌هایی داریم و بعداً چگونه آن‌ها را بررسی می‌کنیم؟»

فرض‌های اولیه‌ام این‌هاست: سابقهٔ موجود برای یک تحلیل مقدماتی کافی باشد تعریف عملیاتی ریزش با رفتار خرید سازگار شود ویژگی‌ها بتوانند تفاوت مشتری‌ها را نشان بدهند و هر مرحلهٔ تکمیلی چیزی به خروجی اضافه کند.

کفایت تاریخچه را با پوشش و تعداد خرید، تعریف ریزش را با تحلیل حساسیت، و مدل‌ها را با آزمون کنارگذاشته‌شده بررسی می‌کنم. نبود شناسه را الزاماً تصادفی فرض نمی‌کنم و فقط تا جایی که داده اجازه دهد اختلاف‌ها را می‌سنجم. اثر فصل، قابل‌تفکیک‌بودن مشتری حجیم و سودمندبودن اقدام هم فرض‌اند اثر علّی اقدام با این فایل قابل‌آزمون نیست و به آزمایش آینده نیاز دارد.


# بخش سوم: تعیین اهداف داده‌کاوی


## سوال ۲۶

«هدف داده‌کاوی این پروژه چیست؟»

می‌خواهم یک خروجی مشتری‌محور بسازم که تیپ رفتاری، امتیاز و سطح خطر و اولویت پیگیری را نشان بدهد و دلیلش قابل‌فهم باشد. اگر داده و ارزیابی پشتیبانی کردند، قواعد پیشنهاد محصول و ارزش درآمدی آینده هم اضافه می‌شود.

پس هدفم صرفاً اجرای پنج الگوریتم نیست. هر بخش باید به این خروجی کمک مشخصی کند وگرنه می‌تواند ساده یا حذف شود.


## سوال ۲۷

«تفاوت هدف کسب‌وکار و هدف داده‌کاوی چیست؟»

هدف کسب‌وکار یک تغییر واقعی است: مثلاً مشتری بیشتری بماند یا منابع حفظ مشتری بهتر مصرف شود. هدف داده‌کاوی یک ابزار تحلیلی است: گروه، امتیاز خطر و فهرست پیگیری.

من در این تمرین می‌توانم کیفیت ابزار را ارزیابی کنم. برای ادعای تغییر واقعی کسب‌وکار باید اقدام اجرا شود و اثر آن جدا اندازه‌گیری شود.


## سوال ۲۸

«هدف داده‌کاوی را در یک جمله بنویسید.»

با استفاده از سابقهٔ در دسترس قبل از زمان مرجع، مدلی قابل‌توضیح بسازم و ارزیابی کنم که مشتری‌های واجد شرایط را از نظر خطر ریزش رتبه‌بندی کند و در کنار گروه رفتاری و ارزش احتمالی، فهرست اولویت‌دار پیگیری بدهد.


## سوال ۲۹

«آیا این پروژه طبقه‌بندی، رگرسیون، خوشه‌بندی، قواعد انجمنی است یا ترکیبی؟ دلایل خود را توضیح دهید.»

در نسخهٔ فعلی، مسئله ترکیبی است. خوشه‌بندی برای شناخت تیپ مشتری، طبقه‌بندی برای خطر، قواعد انجمنی برای محصول مرتبط و مدل احتمالی خرید برای تکمیل ارزش آینده است. بررسی کیفیت داده با قواعد شروع می‌شود و Isolation Forest فقط در صورت فایدهٔ مشخص برای علامت‌زدن رفتار غیرعادی اضافه می‌شود.

انتخاب اولیهٔ الگوریتم‌ها K-Means، CART، Apriori یا FP-Growth و BG/NBD همراه Gamma-Gamma است. نوع بخش‌ها از ایدهٔ خودم آمده و برای مقایسه و انتخاب الگوریتم مشخص از هوش مصنوعی هم مشورت گرفته‌ام بهترین‌بودنشان هنوز ثابت نشده است.

پنج سطح خطر از امتیاز CART می‌آید، نه شناسهٔ خوشه. قاعدهٔ Recency و لجستیک baseline هستند مدل قوی‌تر هم در صورت نیاز برای مقایسه است. BG/NBD فعلاً ورودی طبقه‌بند نیست و خروجی آن را تأیید مستقل ریزش نمی‌دانم.


## سوال ۳۰

«اگر خروجی پروژه احتمال ریزش باشد، نوع مسئله چیست؟»

اگر یک برچسب دوحالتی برای ریزش تعریف کنم و از گذشته احتمال آن را یاد بگیرم، مسئله طبقه‌بندی دودویی است. احتمال درخت خودکار کالیبره نیست باید کیفیت آن را جدا بررسی کنم.

اگر احتمال قابل‌اعتماد نبود، بهتر است خروجی را امتیاز نسبی خطر بنامم. دسته‌بندی همین امتیاز به پنج سطح هم به معنی آموزش یک مدل پنج‌کلاسه نیست.


## سوال ۳۱

«اگر خروجی پروژه شدت کاهش خرید باشد، نوع مسئله چیست؟»

اگر نسبت افت را به‌صورت عدد پیوسته پیش‌بینی کنم، رگرسیون مطرح است. اگر خودِ شدت را با برچسب‌های مرتب تعریف و پیش‌بینی کنم، مسئلهٔ ترتیبی داریم.

ولی فعلاً پنج سطح پروژهٔ من سطح خطر وقوع ریزش است، نه شدت افت. نسبت افت را برای ساخت و بررسی برچسب نگه می‌دارم و بدون نیاز روشن، مدل جدا برای پیش‌بینی شدت به پروژه اضافه نمی‌کنم.


## سوال ۳۲

«اگر خروجی پروژه گروه‌بندی مشتریان باشد، نوع مسئله چیست؟»

اینجا خوشه‌بندی مطرح است، چون برچسب آمادهٔ تیپ مشتری ندارم. می‌خواهم از ویژگی‌هایی مثل تازگی، تعداد و مبلغ خرید، گروه‌های رفتاری بسازم و بعد پروفایلشان را توضیح بدهم.

اسم‌هایی مثل وفادار یا کم‌خرید باید از ویژگی‌های واقعی گروه بیایند. برای نام «رو به افت» هم ویژگی روند لازم است صرف کم‌خریدبودن به معنی افت نیست.


## سوال ۳۳

«چه معیارهایی برای ارزیابی مدل اهمیت دارند و چرا دقت کافی نیست؟»

Accuracy به‌تنهایی کافی نیست حتی بدون عدم‌توازن شدید هم نشان نمی‌دهد بالای فهرست چقدر مفید است. برای خطر، PR-AUC و Precision@K، Recall@K و Lift@K را مهم می‌دانم. برای احتمال، نمودار کالیبراسیون و Brier Score را بررسی می‌کنم و ماتریس خطا هم به درک اشتباه‌ها کمک می‌کند.

برای خوشه‌ها، Silhouette و پروفایل و پایداری مهم‌اند DB و CH هم مکمل‌اند. برای قواعد، support، confidence و lift و برای خرید آینده MAE و مقایسه با baseline لازم است. نرخ واقعی ریزش و شدت عدم‌توازن را هنوز باید با تعریف نهایی محاسبه کنم.


## سوال ۳۴

«با توجه به هزینهٔ مثبت کاذب و منفی کاذب، آستانه کجاست و کدام خطا برای کدام گروه بدتر است؟»

مثبت کاذب یعنی مشتری را پرخطر بدانم ولی طبق تعریف و افق ارزیابی ریزش نکند منفی کاذب یعنی ریزش کند ولی مدل او را شناسایی نکند. در سناریوی اجرا، اولی ممکن است پیگیری بی‌مورد و دومی فرصت ازدست‌رفته ایجاد کند.

این تعریف با «بدون تماس هم می‌ماند» فرق دارد آن یک سؤال علّی دربارهٔ اثر اقدام است. از ماتریس خطا پاسخ آن را نمی‌گیرم.

آستانهٔ ۰٫۵ الزام نیست. ظرفیت پیگیری، هزینه‌ها و ارزش مشتری می‌توانند روی انتخاب اثر بگذارند. چون هزینهٔ واقعی ندارم، چند ظرفیت یا سناریو را روی دادهٔ توسعه مقایسه می‌کنم و آزمون نهایی را برای تنظیم آستانه مصرف نمی‌کنم.


## سوال ۳۵

«حداقل سه معیار فنی برای موفقیت مدل تعریف کنید.»

سه معیار اصلی پیشنهادی من این‌ها هستند: PR-AUC نسبت به baseline، Precision@K و Lift@K در ظرفیت پیگیری مشخص، و بیشترشدن نرخ ریزش واقعی از سطح کم‌خطر به پرخطر روی آزمون. برای احتمال‌ها هم کالیبراسیون را می‌سنجم.

برای خوشه‌ها، معنی‌داربودن پروفایل و پایداری را کنار Silhouette می‌گذارم حداقل اندازهٔ اجباری پنج درصد یا ARI ثابت را قانون عمومی نمی‌دانم. ARI نمونه‌گیری مجدد هم باید روی مجموعهٔ مرجع یا مشتری‌های مشترک محاسبه شود.

مدل خرید آینده باید از یک baseline ساده در MAE بهتر باشد، ولی این معیار اعتبارسنجی کامل CLV نیست. هدف عددی دقیق را با دادهٔ توسعه تعیین و پیش از دیدن آزمون ثبت می‌کنم. جداسازی درست داده‌ها و تکرارپذیری هم شرط پایهٔ پذیرش‌اند، نه جایگزین عملکرد.


## سوال ۳۶

«حداقل سه معیار کسب‌وکاری برای موفقیت پروژه تعریف کنید.»

در اجرای واقعی، حداقل سه معیار می‌تواند افزایش حفظ مشتری نسبت به کنترل، سود افزایشی پس از هزینهٔ اقدام و استفادهٔ واقعی تیم از فهرست باشد. کاهش تماس نامناسب و رضایت مشتری هم مهم است.

در این تمرین دادهٔ لازم برای سنجش این‌ها را ندارم. بنابراین آن‌ها را پیشنهاد ارزیابی آینده می‌نویسم. نقطهٔ سربه‌سر هم باید با حاشیهٔ سود و اثر افزایشی اقدام محاسبه شود بیشترشدن درآمد خام از هزینه به‌تنهایی کافی نیست.


## سوال ۳۷

«یک بیانیهٔ رسمی هدف داده‌کاوی بنویسید.»

«هدف داده‌کاوی، ساخت و اعتبارسنجی یک روش ترکیبی و قابل‌توضیح برای شناخت گروه‌های مشتری، برآورد خطر ریزش و تولید فهرست اولویت‌دار پیگیری است پیشنهاد محصول و برآورد ارزش آینده نیز در صورت کفایت داده، خروجی را تکمیل می‌کنند.»

معیارها و هدف‌های عددی روی بخش توسعه مشخص و قبل از ارزیابی نهایی ثابت می‌شوند.


## سوال ۳۸

«رابطهٔ هدف داده‌کاوی و هدف کسب‌وکار را توضیح دهید.»

من از تصمیم کسب‌وکار شروع کردم و بعد پرسیدم چه تحلیل‌هایی برای آن لازم است. چون تصمیم اصلی اولویت پیگیری است، کیفیت بالای فهرست برایم مهم‌تر از داشتن مدل پیچیده است.

اگر هدف به جذب خرید دوم یا بازگرداندن مشتری غیر فعال تغییر کند، جامعهٔ هدف و برچسب هم باید عوض شود. این ارتباط باعث می‌شود بعد از شناخت داده بتوانم به تعریف مسئله برگردم.


## سوال ۳۹

«چگونه کل خط لولهٔ ترکیبی را به‌صورت سرتاسری اعتبارسنجی می‌کنیم؟»

اول زمان مرجع، مشتری‌های واجد شرایط و افق آینده را مشخص می‌کنم. سپس دادهٔ آموزش، تنظیم و آزمون را جدا می‌کنم خوشه‌بندی، تبدیل‌ها و هر بخش قابل‌آموزش فقط با دادهٔ مجاز همان ارزیابی ساخته می‌شوند. جداسازی زمانی ویژگی و برچسب، جای این جداسازی را نمی‌گیرد.

مدل خطر را با baseline و کیفیت بالای فهرست می‌سنجم خوشه‌ها را از نظر پروفایل و پایداری، قواعد را در دادهٔ کنارگذاشته‌شده و مدل خرید را با رفتار واقعی بازهٔ بعد بررسی می‌کنم. اگر چند نقطهٔ برش داشته باشم، تکرار یک مشتری و هم‌پوشانی زمان‌ها را هم کنترل می‌کنم.

در پایان باید جدول نهایی از نظر سازگاری گروه، خطر، ارزش و توضیح بررسی شود. اثر اقدام روی حفظ مشتری، ارزیابی جداگانهٔ آینده است و جزو نتیجهٔ همین آزمون تاریخی نیست.


## سوال ۴۰

«چگونه بدون گروه کنترل، اثر نگهداشت را اندازه می‌گیریم و مسئلهٔ counterfactual چیست؟»

برای یک مشتری نمی‌توانم هم‌زمان ببینم اگر تماس بگیرم چه می‌شود و اگر تماس نگیرم چه می‌شود. این همان مسئلهٔ وضعیتِ خلاف‌واقع است. ماندن مشتری بعد از پیشنهاد به‌تنهایی اثر پیشنهاد را ثابت نمی‌کند.

آزمایش تصادفی با گروه کنترل راه مناسبی برای برآورد اثر است. روش‌های مشاهده‌ای هم به اطلاعات اقدام، پیامد و متغیرهای مرتبط و فرض‌های قوی نیاز دارند صرف نداشتن گروه کنترل با اجرای uplift حل نمی‌شود. فایل فعلی حتی تاریخچهٔ اقدام ندارد، پس اینجا فقط ارزیابی خطر را انجام می‌دهم.


# بخش چهارم: برنامهٔ پروژه


## سوال ۴۱

«یک ساختار شکست کار و وابستگی بین فازها طراحی کنید.»

کار را به این بسته‌ها می‌شکنم: تثبیت تعریف مسئله ممیزی داده آماده‌سازی تراکنش و مشتری ساخت baseline و ویژگی‌ها بررسی خوشه و مدل خطر قواعد و ارزش آینده ارزیابی و تحویل.

قبل از مدل، تعریف خرید، جامعهٔ هدف و جداسازی داده باید روشن باشد. اگر ویژگی خوشه وارد طبقه‌بند شود، خوشه‌بندی در هر تقسیم فقط روی آموزش برازش می‌شود و بقیه به آن تخصیص داده می‌شوند. اضافه‌کردن ویژگی خوشه هم باید فایده‌اش را نشان بدهد.

قواعد و مدل ارزش بعد از آماده‌شدن دادهٔ مناسب می‌توانند جدا پیش بروند. لازم نیست پنج مرحله را یک زنجیرهٔ اجباری بدانم. منظورم وابستگی کارهاست، نه داشتن چند عضو واقعی در تیم.


## سوال ۴۲

«چه تخصص‌هایی لازم است و اعضای تیم و مسئولیت هر کدام چیست؟»

در یک شرکت واقعی، تحلیلگر کسب‌وکار یا CRM برای تعریف نیاز، دانشمند داده برای مدل، مهندس داده برای آماده‌سازی و مسئول بازاریابی برای اجرا لازم‌اند. مالی، IT، خدمات مشتری و مسئول حریم خصوصی هم بسته به دامنه نقش دارند.

ولی این تمرین را خودم انجام می‌دهم و استاد بازخورد آموزشی می‌دهد چنین تیمی در اختیار ندارم. این فهرست نشان می‌دهد در کاربرد واقعی چه تخصص‌هایی کم دارم، نه اینکه این افراد عضو پروژهٔ من هستند.


## سوال ۴۳

«چه منابع و ابزارهای نرم‌افزاری لازم است؟»

برای شروع، پایتون و ژوپیتر و کتابخانه‌های خواندن و تحلیل داده مثل pandas، ابزارهای مدل‌سازی scikit-learn و در صورت نیاز ابزار قواعد و مدل‌های خرید کافی‌اند. برای BG/NBD و Gamma-Gamma باید ابزار مناسب و قابل‌اجرای محیط را بررسی کنم.

نسخه‌بندی، ثبت پارامترها و امکان اجرای دوباره از ابزار ارائهٔ خاص مهم‌ترند. Power BI یا ابزارهای مشابه اختیاری‌اند برای این تمرین نوت‌بوک و جدول خروجی می‌توانند کافی باشند. فعلاً دلیلی برای تعهد به ابزار شرکتی خاص ندارم.


## سوال ۴۴

«یک زمان‌بندی برای پروژه طراحی کنید و زمان هر فاز را تخمین بزنید.»

چون مهلت دقیق دوره را اینجا ندارم، یک تخمین اولیهٔ قابل‌بازبینی برای کار دانشجویی می‌گذارم: درک کسب‌وکار ۱ تا ۲ روز کاری، درک داده ۲ تا ۳ روز، آماده‌سازی ۳ تا ۵ روز، مدل‌سازی ۴ تا ۶ روز، ارزیابی ۲ تا ۳ روز و تحویل ۱ تا ۲ روز. مجموع حدود ۱۳ تا ۲۱ روز کاری است نه وعدهٔ قطعی زمان تحویل.

حدود ۲۰ درصد زمان اضافه هم برای برگشت و اصلاح در نظر می‌گیرم. این‌ها زمان کار فعال فرضی‌اند و با مهلت واقعی و زمان در دسترسم تنظیم می‌شوند. اگر وقت کم باشد، اول خروجی اصلی خطر و ارزیابی آن را کامل می‌کنم و بخش‌های تکمیلی را محدود می‌کنم.


## سوال ۴۵

«چه خروجی‌هایی باید به مدیریت و چه خروجی‌هایی به بازاریابی و CRM تحویل داده شود؟»

برای مدیریت فرضی، خلاصهٔ مسئله، اندازه و ویژگی گروه‌های در معرض خطر، معیارهای آزمون و محدودیت‌ها را می‌دهم. برای استفاده‌کنندهٔ فهرست، جدول مشتری با گروه، خطر، دلیل و اولویت پیگیری و در صورت اعتبار، پیشنهاد محصول و ارزش آینده را آماده می‌کنم.

در تحویل واقعی دوره، این‌ها به نوت‌بوک‌های مستند، جدول نهایی و گزارش خلاصه تبدیل می‌شوند. نتیجهٔ کمپین یا اثبات درآمد حفظ‌شده جزو خروجی نیست، چون اجرا نشده است.


## سوال ۴۶

«معیار پذیرش پروژه توسط کارفرما چیست؟»

در این تمرین، پذیرش را این‌طور تعریف می‌کنم: تعریف ریزش روشن باشد، همهٔ مشتری‌های وارد یا خارج از امتیازدهی قابل‌شناسایی باشند، ویژگی‌ها فقط از بازهٔ مشاهده ساخته شده باشند، نتیجه روی آزمون و در مقایسه با baseline گزارش شود و جدول نهایی قابل‌توضیح و بازتولید باشد.

در شرکت واقعی، تأیید مالک تصمیم، امکان اجرای اقدام و بررسی هزینه و شرایط استفاده از داده هم لازم است. این تأییدها را در گزارش آموزشی انجام‌شده فرض نمی‌کنم.


## سوال ۴۷

«حداقل پنج ریسک اجرایی پروژه را فهرست کنید و برای هر کدام برنامهٔ مقابله تعریف کنید.»

چند ریسک اجرایی و راه مقابلهٔ پیشنهادی‌ام:

- فایل یا مسیر نامعتبر: مسیر ورودی و نسخهٔ داده را ثبت و بارگذاری را بررسی می‌کنم.
- کیفیت پایین رکوردها: گزارش کیفیت و قواعد پاک‌سازی شفاف می‌سازم.
- بزرگ‌شدن پروژه: خروجی اصلی را از بخش‌های تکمیلی جدا نگه می‌دارم.
- صحت ارزیابی: مرز زمان و تقسیم آموزش، تنظیم و آزمون را از اول مشخص می‌کنم.
- ناپایداری خوشه یا مدل: با baseline و چند اجرای مناسب مقایسه می‌کنم و پیچیدگی را کم می‌کنم.
- کمبود زمان یا مشکل کتابخانه: محیط قابل‌اجرا و گزینهٔ ساده‌تر نگه می‌دارم.
- ضعف توضیح و مستندات: دلیل انتخاب‌ها و محدودیت هر خروجی را هم‌زمان می‌نویسم.

عدم پذیرش تیم و مشکلات اجرای کمپین هم ریسک سناریوی واقعی‌اند، ولی فعلاً تیم یا کمپین واقعی ندارم.


## سوال ۴۸

«برنامهٔ پروژه را به صورت روایی ارائه دهید.»

اول روشن می‌کنم می‌خواهم چه مشتری‌هایی را برای چه افقی در معرض خطر بدانم. بعد داده را می‌شناسم و می‌بینم این تعریف واقعاً قابل‌اجرا هست یا نه. سپس خرید معتبر، ویژگی‌ها و تقسیم داده را می‌سازم و از یک baseline شروع می‌کنم.

بعد خوشه‌بندی و مدل خطر را بررسی می‌کنم و بخش‌های قواعد و ارزش را فقط در صورت فایده اضافه می‌کنم. در ارزیابی می‌پرسم آیا بالای فهرست واقعاً بهتر شده و پنج سطح خطر معنی دارند؟ در پایان، خروجی و محدودیت‌ها را مستند تحویل می‌دهم. اگر پاسخ یک مرحله نامناسب بود، به تعریف یا آماده‌سازی برمی‌گردم اجرای پنج الگوریتم به‌خودی‌خود پایان کار نیست.


## سوال ۴۹

«مهم‌ترین نقاط عطف پروژه چیست و چه عواملی ممکن است باعث شکست آن شود؟»

نقاط عطفم این‌هاست: نسخهٔ روشن تعریف مسئله گزارش کیفیت داده مجموعهٔ ویژگی و تقسیم معتبر اولین baseline مدل خطر ارزیابی‌شده بررسی فایدهٔ بخش‌های مکمل و جدول و گزارش نهایی.

شکست می‌تواند از تعریف غلط ریزش، ارزیابی نامناسب، کمبود مشتری واجد شرایط، تفاوت چرخهٔ خرید، پیچیدگی بیش از حد یا نبود نتیجهٔ بهتر از روش ساده بیاید. اگر یک روش بهتر نشد، آن را پنهان نمی‌کنم نتیجه و تصمیم بعدی را گزارش می‌کنم. نداشتن کمپین کنترل‌شده محدودیت اثبات اثر است، نه شکست خودکار این تمرین تاریخی.


## سوال ۵۰

«چه برنامه‌ای برای پایش، بازآموزی، تحویل و حاکمیت مدل دارید؟»

برای تحویل فعلی، مسیر داده، قواعد پاک‌سازی، نسخهٔ ابزارها، پارامترها، تعریف برچسب و ترتیب اجرای نوت‌بوک‌ها را ثبت می‌کنم. جدول نهایی هم باید تاریخ مرجع و معنی ستون‌ها را داشته باشد.

پایش ماهانه، drift و بازآموزی فقط برای کاربرد آینده پیشنهاد می‌شوند الان جریان دادهٔ جدیدی ندارم. در شرکت واقعی باید مالک مدل، دسترسی و مسئول بازبینی مشخص شود و بعد از رسیدن پیامدهای واقعی، عملکرد و کالیبراسیون دوباره بررسی شوند. بازآموزی هم باید دلیل داشته باشد، نه صرفاً یک برنامهٔ بی‌ارتباط با تغییر داده.


## سوال ۵۱

«یک خلاصهٔ اجرایی برای مدیرعامل بنویسید.»

در این پروژه می‌خواهم از تاریخچهٔ خرید، مشتری‌هایی را که طبق یک تعریف روشن بیشتر در معرض توقف یا افت خرید هستند شناسایی کنم و فهرستی قابل‌توضیح برای اولویت پیگیری بسازم. طرح پیشنهادی من شناخت گروه‌های رفتاری، برآورد خطر، پیشنهاد محصول مرتبط و در صورت اعتبار، تخمین ارزش درآمدی آینده است.

انتخاب اولیهٔ الگوریتم‌ها با بررسی و مشورت انجام شده، اما فایدهٔ هر بخش باید در اجرا ثابت شود. مهم‌ترین محدودیت‌ها، تاریخچهٔ کوتاه، شناسهٔ خالی، فصل خرید و نبود اطلاعات سود و کمپین‌اند. خروجی این تمرین یک تحلیل و پیشنهاد قابل‌بررسی است کاهش واقعی ریزش یا سود کمپین فقط در اجرای بعدی با اندازه‌گیری اثر قابل‌اثبات خواهد بود.


## جمع‌بندی این فاز

الان هدف، خروجی، دامنه و معیارهای پیشنهادی روشن‌تر شده‌اند. تعریف‌ها و آستانه‌ها هنوز باید در فاز درک داده محک بخورند. این سند را یک نقشهٔ قابل‌اصلاح می‌دانم اگر داده با فرضم نخواند، فرض یا روش را اصلاح می‌کنم.
